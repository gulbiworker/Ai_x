```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("C:\Ai_x\lecNote\FlowCast\Data_Analysis\dat.csv")  


data = df[['rental_count', 'total_population']].dropna()


plt.figure(figsize=(12, 4))

for i, col in enumerate(data.columns):
    plt.subplot(1, 2, i + 1)
    sns.violinplot(y=data[col])
    plt.title(f"{col} 분포 (Violin Plot)")
    plt.grid(False)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis


df = pd.read_csv("C:\Ai_x\lecNote\FlowCast\Data_Analysis\dat.csv")  # 예: 'data.csv'


data = df[['rental_count', 'total_population']].dropna()


summary = {
    '변수명': [],
    '왜도 (Skewness)': [],
    '첨도 (Kurtosis)': []
}

for col in data.columns:
    summary['변수명'].append(col)
    summary['왜도 (Skewness)'].append(round(skew(data[col]), 3))
    summary['첨도 (Kurtosis)'].append(round(kurtosis(data[col]), 3))

summary_df = pd.DataFrame(summary)


plt.figure(figsize=(12, 4))

for i, col in enumerate(data.columns):
    plt.subplot(1, 2, i+1)
    sns.histplot(data[col], bins=30, kde=True)
    skew_val = summary['왜도 (Skewness)'][i]
    kurt_val = summary['첨도 (Kurtosis)'][i]
    plt.title(f"{col}\n왜도={skew_val}, 첨도={kurt_val}")
    plt.xlabel(col)
    plt.ylabel("빈도수")
    plt.grid(False)

plt.tight_layout()
plt.show()

# 📋 표로 출력
display(summary_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 불러오기
df = pd.read_csv(r"C:\Ai_x\lecNote\FlowCast\Data_Analysis\dat.csv")

# rental_count 500 이상 필터링
filtered = df[df['rental_count'] >= 500]['rental_count'].dropna()

# 히스토그램 (격자선 제거)
plt.figure(figsize=(6, 4))
sns.histplot(filtered, bins=30, kde=True)
plt.title("대여량 500 이상만 추출한 히스토그램")
plt.xlabel("rental_count")
plt.ylabel("빈도수")
plt.grid(False)  # ✅ 격자선 끄기
plt.tight_layout()
plt.show()

# 바이올린 플롯 (격자선 없음)
plt.figure(figsize=(4, 6))
sns.violinplot(y=filtered)
plt.title("대여량 500 이상만 추출한 바이올린 플롯")
plt.grid(False)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 📌 CSV 파일 불러오기
df = pd.read_csv("C:\Ai_x\lecNote\FlowCast\Data_Analysis\dat.csv")  # 예: 'data.csv'

# 📌 기후 변수 목록
climate_vars = ['temperature', 'humidity', 'precipitation', 'wind_speed']

# 📊 히스토그램 그리기 (KDE 없이)
plt.figure(figsize=(12, 8))

for i, col in enumerate(climate_vars):
    plt.subplot(2, 2, i + 1)
    sns.histplot(df[col].dropna(), bins=30, kde=False, edgecolor='black')
    plt.title(f"{col} 히스토그램")
    plt.xlabel(col)
    plt.ylabel("빈도수")
    plt.grid(False)

plt.tight_layout()
plt.show()

```
