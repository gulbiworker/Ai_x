```python
import pandas as pd
import matplotlib.pyplot as plt

# ===== 한글 폰트 설정 =====
plt.rc('font', family='Malgun Gothic')  # 윈도우: 맑은 고딕
plt.rc('axes', unicode_minus=False)     # 마이너스 깨짐 방지

# ===== 데이터 로드 =====
df = pd.read_csv(r"C:\Users\Admin\OneDrive\바탕 화면\인사평가ver2\인사평가_환율적용.csv")  # 경로 수정 필요

# ===== 시각화 대상 컬럼 =====
cols = ["부서", "학력", "전공", "직급", "업무평가"]

# ===== 파이차트 출력 =====
fig, axes = plt.subplots(1, 5, figsize=(25, 5))

for ax, col in zip(axes, cols):
    df[col].value_counts().plot.pie(
        autopct="%.1f%%",
        startangle=90,
        ax=ax,
        textprops={"fontsize": 10}
    )
    ax.set_ylabel("")
    ax.set_title(col)

plt.tight_layout()
plt.show()

# ===== 이미지 저장 =====
plt.savefig("pie_charts.png", dpi=300)  # PNG 파일 저장 (고해상도)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# ===== 한글 폰트 설정 =====
plt.rc('font', family='Malgun Gothic')  # 윈도우: 맑은 고딕
plt.rc('axes', unicode_minus=False)

# ===== 데이터 로드 =====
file_path = r"C:\Users\Admin\OneDrive\바탕 화면\인사평가ver2\인사평가_환율적용.csv"
df = pd.read_csv(file_path)

# ===== 저장 경로 (CSV와 같은 폴더) =====
save_dir = os.path.dirname(file_path)

# ===== 시각화 대상 컬럼 =====
cols = ["부서", "학력", "전공", "직급", "업무평가"]

# ===== 각 컬럼별 파이차트 개별 저장 =====
for col in cols:
    plt.figure(figsize=(6, 6))
    df[col].value_counts().plot.pie(
        autopct="%.1f%%",
        startangle=90,
        textprops={"fontsize": 15}
    )
    plt.ylabel("")
    plt.title(col)
    
    save_path = os.path.join(save_dir, f"{col}_pie.png")
    plt.savefig(save_path, dpi=300)
    plt.close()  # 창 닫기 (메모리 절약)

print("파이차트 이미지 저장 완료!")
```

```python
import matplotlib.pyplot as plt

col = "직급"

plt.figure(figsize=(7, 7))
values = df[col].value_counts()

# 파이차트 (레이블 제거)
values.plot.pie(
    labels=None,   # 조각 안에는 레이블 표시하지 않음
    autopct="%.1f%%",
    startangle=90,
    textprops={"fontsize": 10}
)

plt.ylabel("")
plt.title(col)

# 범례 추가 (오른쪽에 표시)
plt.legend(
    values.index,
    title=col,
    loc="center left",
    bbox_to_anchor=(1, 0.5),
    fontsize=9
)

plt.tight_layout()
plt.show()
```
