```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()  # 오타: load_doten() → load_dotenv()

client = OpenAI()  # 오타: open.OpenAI() → OpenAI()
```

```python
# 변환할 텍스트
text = "bello dochipachi dochipachipochi"
```

```python
response = client.audio.speech.create(
    model = 'tts-1',
    voice = "nova",
    input= text)
```

```python
audio_data = response.content
with open("data/ch5_nova.mp3","wb") as audio_file:
    audio_file.write(audio_data)
```

```python
text = "bsdfasdf asdfa sdfawegadv  asdwehsadf xz xASDFASDGABFGAGCB ADSFEwrasDfasdf "
response_diff_voice = client.audio.speech.create(
    model = 'tts-1',
    voice = "fable",
    input = text,
    speed = 0.9)
audio_data = response_diff_voice.content
with open("data/ch5_fable.mp3","wb") as audio_file:
    audio_file.write(audio_data)
```

```python
# 파일 삭제
import os
try:
    os.remove("ch5_nova.mp3")
    print("파일 삭제 완료")
except Exception as e:
    print('파일 삭제 중 오류 발생 :', e)
```

```python
# 1. 음성파일 열기
audio_file = open('data/ch5_fable.mp3', 'rb')

response = client.audio.transcriptions.create(
    file=audio_file,
    model = 'whisper-1',
    response_format='text')

print(response)
audio_file.close()
```
