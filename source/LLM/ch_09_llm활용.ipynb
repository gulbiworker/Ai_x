# 4. upstage 의 embedding model 사용

```python
import numpy as np
def cosine_similarity(vec1, vec2):
    """두 백터 사이의 코사인 유사도 계산"""
    dot_product = np.dot(vec1, vec2)
    norm_vec1 = np.linalg.norm(vec1) # 벡터의 길이
    norm_vec2 = np.linalg.norm(vec2)
    if norm_vec1==0 or norm_vec2==0:
        return 0.0
    return dot_product / (norm_vec1*norm_vec2)
```

# 쪼갠 문서를 임베딩 -> 벡터 데이터베이스 저장

```python
from dotenv import load_dotenv
from langchain_openai import OpenAIEmbeddings
load_dotenv()
# https://python.langchain.com/v0.2/docs/how_to/embed_text/
embedding = OpenAIEmbeddings(
    model = "text-embedding-3-large"
)
embeddings = embedding.embed_documents(
    [
        "소득세법 어쩌구 저쩌구",
        document[0].page_content
    ]
)
len(embeddings), len(embeddings[0])
```

```python
embeddings = embedding.embed_documents(
    [
        "소득세법 어쩌구 저쩌구",
        document[0].page_content
    ]
)
len(embeddings), len(embeddings[0])
```

```python
%%time
from langchain_chroma import Chroma
database = Chroma.from_documents(
    documents=document,
    embedding=embedding,
    collection_name="tax-collection", # 생략시 이름 랜덤
    persist_directory='./chroma'
)
```

```python
query = "연봉 1000만원인 직장인의 소득세는 얼마인가"
retrieved_docs = database.similarity_search(query,
                                           k=3)
```

# 유사도 검색으로 가져온 문서를 질문과 같이 llm 전달하여 답변 생성

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model='gpt-4.1-nano')
```

```python
prompt = f"""[identity]
- 당신은 아주 잔인아고 야만적인 언어를 사용하는 오크족장이다
- [context]를 참고해서 사용자의 질문에 답변해 주세요
[context]는 다음과 같아요
{retrieved_docs}
Question : {query}"""
```

```python
ai_message = llm.invoke(prompt)
```

```python
print(ai_message.content)
```

```python
from langchain.prompts import PromptTemplate

# 문자열을 PromptTemplate으로 변환
prompt_template = PromptTemplate(
    input_variables=["context", "question"],
    template="""
    [identity]
    - 당신은 친절하고 똑똑한 AI입니다.
    
    [context]
    {context}
    
    [question]
    {question}
    
    [instruction]
    위 내용을 바탕으로 정답을 알려주세요.
    """
)

# 수정된 코드
from langchain.chains import RetrievalQA

qa_chain = RetrievalQA.from_chain_type(
    llm,
    retriever=database.as_retriever(search_kwargs={"k": 5}),
    chain_type_kwargs={"prompt": prompt_template}
)
```
