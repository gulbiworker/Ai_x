```python
# 웹 예제
def askGpt(prompt):
    "gpt에게 prompt여청 결과 반환"
    client = OpenAI()
    response = client.chat.completions.create(
    model="gpt-4.1-nano",
    message=[{"role":"user", "content":"당신은 텍스트를 한국어 요약하는 전문 어시스턴스입니다"},
             {"role":'user','content':prompt}])
    return response.choices[0].messafe.content
```

```python
# 스트리밍 예제 : 문장을 한글자씩 받아 출력
import time
message = [
    {"role":"system", 
     "content":"대한민국을 사랑하는 도우미입니다. 도시 이름 한글자씩 출력하는 도우미입니다. 다른 문장 금지입니다"},
    {"role":"user", "content":"아시아 국가들의 수도를 5개 알려줘. 도시이름만 출력해 줘"}
]
response_stream = client.chat.completions.create(
    model="gpt-4.1-nano",
    messages=message,
    stream=True # 스트리링으로 응답
)
print("실시간 응답 :", end="")
for chunk in response_stream:
    # 스트리밍으로 들어온 조각에서 추가된 content부분 추출
    chunk_message = chunk.choices[0].delta.content
    if chunk_message is not None:
        print(chunk_message, end=" / ")
        time.sleep(0.5)
```

```python
from dotenv import load_dotenv
from openai import OpenAI, OpenAIError
load_dotenv()
client = OpenAI()
```

```python
# 대화 이력을 저장할 리스트 초기화
chat_history = [
    {"role": "system", "content": "당신은 AI 지식이 풍부한 상담원 그리고 반지의 제왕 오크처럼 말함"}
]

print("챗봇 시작! 종료하려면: exit, quit, bye, 종료")

while True:
    user_input = input("사용자: ").strip()

    # 종료 조건 확인
    if user_input.lower() in ["exit", "quit", "bye", "종료"]:
        print("종료")
        break

    # 빈 입력은 무시
    if user_input == "":
        continue

    # 사용자 입력을 대화 목록에 추가
    chat_history.append({"role": "user", "content": user_input})

    try:
        # OpenAI GPT 호출
        response = client.chat.completions.create(
            model='gpt-4.1-nano',  # 실제 존재하는 모델 이름으로 수정 필요
            messages=chat_history  # 오타: message → messages
        )

        # AI의 응답을 받아서 출력하고 대화에 추가
        ai_message = response.choices[0].message.content
        print("AI:", ai_message)
        chat_history.append({"role": "assistant", "content": ai_message})

    except OpenAIError as e:
        print("오류가 발생했습니다:", e)
        continue
    assistant_reply = response.choices[0].message.content.strip()
    print("ai 답:", assistant_reply)
    chat_history.append(
        {"role": "assistant", "content": assitant_reply})
```
