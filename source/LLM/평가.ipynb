다음의 생성형 ai를 렝체인으로 구현하시오.
▪ 원하는 모델을 이용하여 나라를 입력하면 해당 나라의 가장 유명한 음식을 추천하는 렝체인을 구현
(food_chain))
▪ 음식을 입력하면 레시피를 생성하는 렝체인을 구현(recipe_chain)
▪ 위의 두 체인을 연결하여, 나라이름을 입력받아, 해당 나라의 가장 유명한 음식의 레시피를 생성하는
렝체인 프로그램을 구현하시오.

```python
from langchain_community.chat_models import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="llama3.2:1b")  # 모델 이름은 실제 설치된 ollama 모델 기준

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant with expertise in food."),
    ("human", "What is the most famous food of {country}? Return the name of the food only.")
])

output_parser = StrOutputParser()

output_parser.invoke(llm.invoke(chat_prompt_template.invoke({"country":"Korea"})))
```

```python
from langchain_core.prompts import PromptTemplate
country_prompt = PromptTemplate(
    template = """Guess the receipe of the country based on the following information:
    {information}
    Return the name of the country only""",
    input_variables=['information'])

output_parser.invoke(llm.invoke(country_prompt.invoke({"information":
                                                      "This country is very famous for its wine"})))
```

```python
from langchain_community.chat_models import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="llama3.2:1b", temperature=0)

food_prompt = PromptTemplate(
    template="What is the most famous food of {country}? Return only the name of the food.",
    input_variables=["country"]
)
food_chain = food_prompt | llm | StrOutputParser()
```

```python

recipe_prompt = PromptTemplate(
    template="Give me a simple recipe for {food}. W.",
    input_variables=["food"]
)
recipe_chain = recipe_prompt | llm | StrOutputParser()
```

```python
from langchain_core.runnables import RunnableLambda

full_chain = food_chain | RunnableLambda(lambda food: {"food": food}) | recipe_chain
result = full_chain.invoke({"country": "Korea"})
print(result)
```
