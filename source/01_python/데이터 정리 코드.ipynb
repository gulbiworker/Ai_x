```python
import os
import pandas as pd
from dotenv import load_dotenv

load_dotenv()

# 경로를 raw string 형식으로 지정
input_folder = r"C:\Users\Admin\Desktop\프로젝트\24년도"
output_folder = r"C:\Users\Admin\Desktop\프로젝트\24년도\24년 최종"

# 출력 폴더가 없으면 생성
os.makedirs(output_folder, exist_ok=True)

# 모든 .csv 파일 처리
for filename in os.listdir(input_folder):
    if filename.endswith(".csv"):
        input_path = os.path.join(input_folder, filename)
        output_path = os.path.join(output_folder, filename)

        # 인코딩 시도
        try:
            df = pd.read_csv(input_path, encoding='utf-8')
        except UnicodeDecodeError:
            try:
                df = pd.read_csv(input_path, encoding='euc-kr')
            except:
                df = pd.read_csv(input_path, encoding='cp949')

        # "Start :" 텍스트 포함된 행 제거
        df_cleaned = df[~df.apply(lambda row: row.astype(str).str.contains("Start :").any(), axis=1)]

        # 저장
        df_cleaned.to_csv(output_path, index=False, encoding='utf-8-sig')

print("✅ 모든 파일 처리 완료!")

```
