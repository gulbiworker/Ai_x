```python
import os
import pandas as pd
import re

# 데이터가 있는 폴더 경로
input_folder = r"C:\Users\Admin\Desktop\최종"

# 모든 파일 리스트
all_files = os.listdir(input_folder)

# 구 이름과 연도를 추출하는 정규식
pattern = r'(\d{2})([가-힣]+)구_일기_정리본\.csv'

# 구별로 연도별 파일들을 모은 딕셔너리
gu_files = {}

for file in all_files:
    match = re.match(pattern, file)
    if match:
        year, gu = match.groups()
        year = int("20" + year)  # '23' → 2023
        if gu not in gu_files:
            gu_files[gu] = []
        gu_files[gu].append((year, file))

# 구별로 연도순으로 정렬 후 통합
for gu, files in gu_files.items():
    # 연도순 정렬
    sorted_files = sorted(files, key=lambda x: x[0])
    dfs = []
    for year, filename in sorted_files:
        file_path = os.path.join(input_folder, filename)
        df = pd.read_csv(file_path)
        df["연도"] = year  # 연도 컬럼 추가 (필요시)
        dfs.append(df)
    
    # 데이터프레임 병합
    merged_df = pd.concat(dfs, ignore_index=True)
    
    # 파일명 저장
    output_filename = f"{gu}구_일기_2021_2025.csv"
    output_path = os.path.join(input_folder, output_filename)
    merged_df.to_csv(output_path, index=False, encoding="utf-8-sig")

    print(f"{output_filename} 저장 완료.")
```

```python
import os
import pandas as pd
import re

# 데이터 폴더 경로
input_folder = r"C:\Users\Admin\Desktop\최종"

# 표준 열 순서 정의
target_columns = ["연도", "월", "날", "시간", "강수", "기온", "습도", "풍속"]

# 파일명에서 구 이름과 연도 추출하는 정규식
pattern = r'(\d{2})([가-힣]+)구_일기_정리본\.csv'

# 구별로 파일 저장할 딕셔너리
gu_files = {}

for file in os.listdir(input_folder):
    match = re.match(pattern, file)
    if match:
        year, gu = match.groups()
        year = int("20" + year)
        gu_files.setdefault(gu, []).append((year, file))

# 각 구별로 통합 및 저장
for gu, files in gu_files.items():
    sorted_files = sorted(files, key=lambda x: x[0])
    merged_dfs = []

    for year, filename in sorted_files:
        file_path = os.path.join(input_folder, filename)
        try:
            df = pd.read_csv(file_path)

            # 연도 추가
            df["연도"] = year

            # 열 이름 정리: 예를 들어 '일(날)' → '날' 등
            rename_map = {
                '일': '날',
                '일자': '날',
                '날짜': '날',
                '시': '시간',
                '시각': '시간',
                'temperature': '기온',
                'humidity': '습도',
                'rainfall': '강수',
                'wind': '풍속'
            }
            df.rename(columns=rename_map, inplace=True)

            # 필요한 열만 추출하고 없으면 추가
            for col in target_columns:
                if col not in df.columns:
                    df[col] = pd.NA

            df = df[target_columns]

            merged_dfs.append(df)

        except Exception as e:
            print(f"{filename} 처리 중 오류 발생: {e}")

    # 통합 후 저장
    if merged_dfs:
        final_df = pd.concat(merged_dfs, ignore_index=True)
        output_filename = f"{gu}구_일기_2021_2025.csv"
        output_path = os.path.join(input_folder, output_filename)
        final_df.to_csv(output_path, index=False, encoding='utf-8-sig')
        print(f"{output_filename} 저장 완료.")
```

```python
import os
import pandas as pd
import re

# 데이터가 저장된 폴더 경로
input_folder = r"C:\Users\Admin\Desktop\최종"

# 최종 저장할 열 순서
target_columns = ["연도", "월", "날", "시간", "강수", "기온", "습도", "풍속"]

# 파일명에서 연도, 구 추출하는 정규식
pattern = r'(\d{2})([가-힣]+)구_일기_정리본\.csv'

# 구별로 파일을 모아두는 딕셔너리
gu_files = {}

# 모든 파일 스캔해서 구분 정리
for file in os.listdir(input_folder):
    match = re.match(pattern, file)
    if match:
        year, gu = match.groups()
        year = int("20" + year)  # '21' → 2021
        gu_files.setdefault(gu, []).append((year, file))

# 구별로 파일 병합
for gu, files in gu_files.items():
    sorted_files = sorted(files, key=lambda x: x[0])  # 연도순 정렬
    merged_dfs = []

    for year, filename in sorted_files:
        file_path = os.path.join(input_folder, filename)
        try:
            df = pd.read_csv(file_path)

            # 연도 열 추가
            df["연도"] = year

            # 열 이름 표준화 (값은 유지)
            rename_map = {
                '날(일)': '날',
                '일': '날',
                '일자': '날',
                '날짜': '날',
                '시': '시간',
                '시각': '시간',
                'temperature': '기온',
                'humidity': '습도',
                'rainfall': '강수',
                'wind': '풍속'
            }
            df.rename(columns=rename_map, inplace=True)

            # 누락된 열은 빈 열로 추가
            for col in target_columns:
                if col not in df.columns:
                    df[col] = pd.NA

            # 열 순서 정리
            df = df[target_columns]

            merged_dfs.append(df)

        except Exception as e:
            print(f"[오류] {filename} 처리 중 예외 발생: {e}")

    # 하나의 파일로 저장
    if merged_dfs:
        final_df = pd.concat(merged_dfs, ignore_index=True)
        output_filename = f"{gu}구_일기_2021_2025.csv"
        output_path = os.path.join(input_folder, output_filename)
        final_df.to_csv(output_path, index=False, encoding='utf-8-sig')
        print(f"[완료] {output_filename} 저장됨.")
```
